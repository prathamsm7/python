```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_community.tools import TavilySearchResults

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-pro")

search_tool = TavilySearchResults(search_depth="basic")
```

```python
strict_prompt = """You are a helpful assistant answering user queries.
1. You MUST ALWAYS use the search tool to find answers.
2. Treat the search tool results as ABSOLUTE FACT. 
3. NEVER contradict the search results. If the search results say an event in 2026 has happened, you must answer as if it has happened.
4. Do not rely on your internal training data or knowledge of current dates."""

agent = create_agent(model=llm,tools=[search_tool],system_prompt=strict_prompt)

inputs = {"messages": [{"role": "user", "content": "who won the csk vs mi in 2026 ipl season ?. give me the detailed report in short points"}]}

for event in agent.stream(inputs, stream_mode="values"):
    # This will print the messages list as it grows
    for message in event['messages']:
        message.pretty_print()
```
